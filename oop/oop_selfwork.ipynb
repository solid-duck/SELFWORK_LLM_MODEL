```python
#1 import ABC
from abc import ABC, abstractmethod


#2 creazione classe Person
class Person(ABC):
    def __init__(self, nome, cognome, età):
        self.nome = nome
        self.cognome = cognome
        self.età = età

    @abstractmethod
    def descrizione(self):
        pass


#3 creazione classe Dipendente
class Dipendente(Person):
    def __init__(self, nome, cognome, età, stipendio):
        super().__init__(nome, cognome, età)
        self.stipendio = stipendio

    def descrizione(self):
        return f"{self.nome} {self.cognome}, ha {self.età} anni ed ha un stipendio di {self.stipendio} €."


#4 creazione classe Studente
class Studente(Person):
    def __init__(self, nome, cognome, età, media_scolastica):
        super().__init__(nome, cognome, età)
        self.media_scolastica = media_scolastica

    def descrizione(self):
        return (
            f"{self.nome} {self.cognome} ha {self.età} anni, "
            f"con una media di {self.media_scolastica}."
        )


#5 creazione classe Docente
class Docente(Dipendente):
    def __init__(self, nome, cognome, età, stipendio, materia):
        super().__init__(nome, cognome, età, stipendio)
        self.materia = materia

    def descrizione(self):
        return (
            f"{self.nome} {self.cognome} ha {self.età} anni e ha un stipendio di {self.stipendio} €, è un docente di {self.materia}."
        )


#6 creazione classe Freelancer
class Freelancer(Dipendente):
    def __init__(self, nome, cognome, età, stipendio, partita_iva):
        super().__init__(nome, cognome, età, stipendio)
        self.partita_iva = partita_iva

    def descrizione(self):
        return (
            f"{self.nome} {self.cognome} ha {self.età} anni ed è un freelancer "
            f"con partita IVA {self.partita_iva}."
        )


#7 creazione oggetto Studente
studente = Studente("Luca", "Rossi", 21, 28.5)


#8 creazione oggetto Docente
docente = Docente("Mario", "Verdi", 45, 2200, "Python")


#9 creazione oggetto Freelancer
freelancer = Freelancer("Anna", "Bianchi", 33, 3000, "IT12345678901")


#10 stampa informazioni
print(studente.descrizione())
print(docente.descrizione())
print(freelancer.descrizione())
```
