```python

# TODO Inizializza il client di Open AI
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')

client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=os.getenv("GROQ_API_KEY")
)

```

```python
input_text = input("Enter text: ")

completion = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "system",
            "content": """Sei un analista dati esperto nell'analisi del sentiment dei clienti. Analizza il testo fornito dall'utente e classifica il sentiment in Positivo, Negativo o Neutro.

Formatta l'output esattamente così:
- Sentiment: [Positivo / Negativo / Neutro]
- Motivazione: [Breve spiegazione in 1 frase]
- Parole Chiave: [parola1, parola2, parola3]"""
        },
        {
            "role": "user",
            "content": input_text
        }
    ]
)
#"Ho acquistato queste cuffie la settimana scorsa: il suono è fantastico e la batteria dura tantissimo, ma la spedizione ha ritardato di tre giorni."

print("\n--- RISULTATO DELL' ANALISI")
print(completion.choices[0].message.content)
```
